```python
# def transform_to_gnn_samples(central_user_id, X, y, graph):
Xy = X.copy()
Xy["label"] = y

neighborhood_ids = [central_user_id] + neighbor_ids.values.tolist()
neighborhood_ids = [str(ni) for ni in neighborhood_ids]
```

```python

# 1. Get nodes matching those values
matching_nodes_map = {}
for node, attrs in G.nodes(data=True):
    if attrs.get("twid") in neighborhood_ids:
        matching_nodes_map[attrs["twid"]] = node
```

```python
set(neighborhood_ids).intersection(set(all_user_ids))
```

```python
len(matching_nodes_map)
```

```python

neighborhood_nodes = [matching_nodes_map[ni] for ni in neighborhood_ids]
for i, ni in enumerate(neighborhood_nodes):
    G.nodes[ni]["local_ind"] = i

# 2. Get all edges between those nodes
node_set = set(neighborhood_nodes)  # O(1) lookup
subgraph_edges = [
    (G.nodes[u]["local_ind"], G.nodes[v]["local_ind"]) for u, v in G.edges(neighborhood_nodes)
    if u in node_set and v in node_set
]
```
